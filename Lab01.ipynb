```python
uniform = np.random.rand(100000)

normal = np.random.randn(100000)

expo = np.random.exponential(size=100000)
```

```python
plt.figure(figsize=(10,6))

plt.hist(uniform,
         density=True,
         bins=100,
         histtype="step",
         label="Uniform")

plt.hist(normal,
         density=True,
         bins=100,
         histtype="step",
         label="Normal")

plt.hist(expo,
         density=True,
         bins=100,
         histtype="step",
         label="Exponential")

plt.legend()

plt.title("Random Distributions")

plt.xlabel("Value")

plt.ylabel("Density")

plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D

x = np.linspace(-5,5,100)

y = np.linspace(-5,5,100)

X,Y = np.meshgrid(x,y)

Z = X**2 + Y**2

fig = plt.figure(figsize=(8,6))

ax = fig.add_subplot(111,projection='3d')

ax.plot_surface(X,Y,Z,cmap='viridis')

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.show()
```

```python
df = pd.read_csv("Au_nanoparticle_dataset.csv")
```

```python
new_df = df[
    ["N_total",
     "N_bulk",
     "N_surface",
     "R_avg"]
]
```

```python
new_df.head(20)
```

```python
new_df.describe()
```

```python
fig, axes = plt.subplots(1,4,figsize=(18,4))

for i,col in enumerate(new_df.columns):

    axes[i].hist(new_df[col],bins=30)

    axes[i].set_title(col)

plt.tight_layout()

plt.show()
```

```python
sns.pairplot(new_df)

plt.show()
```

```python
g = sns.PairGrid(new_df)

g.map_upper(sns.histplot)

g.map_diag(sns.histplot, kde=True)

g.map_lower(sns.kdeplot)

plt.show()
```
